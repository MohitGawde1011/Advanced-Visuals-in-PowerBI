1. What is the difference between a Basic Filter, Advanced Filter, and a Slicer in Power BI?
-   | Feature          | Basic Filter                  | Advanced Filter          | Slicer                     |
| ---------------- | ----------------------------- | ------------------------ | -------------------------- |
| Purpose          | Select specific values        | Apply conditions to data | Interactive filtering      |
| Example          | Select 2023                   | Sales > ₹50,000          | Select Region              |
| Location         | Filters pane                  | Filters pane             | Report canvas              |
| User interaction | Limited                       | Condition-based          | Highly interactive         |
| Visibility       | Usually not visible on report | Usually not visible      | Visible directly on report |



Basic Filter: Allows you to select one or more specific values.

Advanced Filter: Allows conditions such as is greater than, contains, starts with, etc.

Slicer: A visual on the report page that allows users to interactively filter other visuals.


2. Apply a Basic Filter to display only sales data for 2023
-   Steps:
Open Sales_dataset in Power BI.
Go to Report View.
Create a table or chart containing the sales data.
Open the Filters pane.
Drag Order Date/Date into Filters on this page or Filters on this visual.
Select 2023 from the date hierarchy.
Apply the filter.
Result:

The visual will display only sales transactions from 2023.

3. Create a Pie Chart of Sales by Product and add a Region Slicer
-   Pie Chart:
Select Pie Chart from the Visualizations pane.
Drag Product into Legend.
Drag Sales into Values.
Make sure Sales is set to Sum.
Rename the title to Sales by Product.
Add Region Slicer:
Select Slicer from the Visualizations pane.
Drag Region into the slicer's Field.
Select a region from the slicer.
Result:

When a user selects a region, the Pie Chart automatically updates to show sales for that region.

4. What is a Visual-Level Filter, and how does it differ from Page-Level and -    Report-Level Filters?
Visual-Level Filter

A Visual-Level Filter affects only one particular visual.

Example:
A bar chart displays sales only for the Technology category, while other visuals remain unchanged.

Page-Level Filter

A Page-Level Filter affects all visuals on the current report page.

Example:
Filter the page to show sales only for 2023.

Report-Level Filter

A Report-Level Filter affects all visuals on all pages of the Power BI report.

Example:
Filter the entire report to show sales only from India.

Difference:

Visual Level → One visual

Page Level → All visuals on one page

Report Level → All visuals on all pages

5. Apply a Report-Level Filter for India
-  Steps:
Open the Filters pane.
Find Filters on all pages.
Drag Country into this section.
Select India.
Apply the filter.
Result:

All visuals across every page of the report will now display sales data only for India.

6. What is the role of Drill-Down in Power BI?

-  Drill-Down allows users to move from a higher-level summary to more detailed levels of data within the same visual.

It helps users explore data without creating separate charts for every level.

Example:

Country → Region → Product Category

A user can start with total sales by Country, click on India, and then see sales by Region, followed by Product Category.

Benefits:
Provides detailed analysis.
Makes reports interactive.
Helps identify trends and problems.
Allows users to move from summary to detailed information.


7. Create a Drill-Down Hierarchy

-   Required hierarchy:

Country → Region → Product Category

Steps:
Select a chart such as a Column Chart.
Drag Country to the X-axis.
Drag Region below Country on the X-axis.
Drag Product Category below Region.
Power BI will create multiple levels in the visual.
Turn on the Drill Down option using the drill-down arrow on the visual.
Click a country such as India.
The chart will move to the Region level.
Drill down again to see Product Category.
Hierarchy:

Country
↓
Region
↓
Product Category

8. Implement Drill-Through for Product

-   Drill-through allows users to select a particular product and navigate to a separate detailed report page.

Step 1: Create a Drill-Through Page
Create a new report page.
Rename it Product Details.
In the Filters pane, locate Drill-through.
Drag Product into the Drill-through field.
Step 2: Add Region-wise Sales

On the Product Details page:

Add a Column Chart.
Drag Region to the X-axis.
Drag Sales to the Y-axis.
Add other useful details such as Quantity or Profit if available.
Step 3: Use Drill-Through
Go back to the main Product visual.
Right-click on a particular Product.
Select Drill-through → Product Details.
Result:

Power BI opens the Product Details page and displays region-wise sales for the selected product.

9. Add a Slicer for Product Category
-  Steps:
Select the Slicer visual.
Drag Product Category into the slicer.
Select a category, such as Electronics.
Observe the drill-down chart.
Result:

The slicer filters the data according to the selected Product Category. The drill-down hierarchy will also reflect the selected filter.

For example:

Country → Region → Product Category

If Electronics is selected, the chart will show only sales related to Electronics.

10. Difference between Drill-Down and Drill-Through

-   | Drill-Down                                           | Drill-Through                                         |
| ---------------------------------------------------- | ----------------------------------------------------- |
| Moves from summary to detail within the same visual. | Opens a separate detailed report page.                |
| Uses a hierarchy.                                    | Uses a drill-through field.                           |
| Example: Country → Region → Product Category.        | Example: Product → Product Details page.              |
| Keeps the user in the same visual/page.              | Takes the user to another report page.                |
| Useful for exploring different levels of data.       | Useful for detailed investigation of a selected item. |
